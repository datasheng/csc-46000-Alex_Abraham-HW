````python
# Start your code here!
import os
import pandas as pd
from openai import OpenAI

# Instantiate an API client
client = OpenAI()

# Continue coding here
# Use as many cells as you like
nasdaq100_ca = pd.read_csv("nasdaq100_CA.csv")
price_change = pd.read_csv("nasdaq100_price_change.csv")

nasdaq100_ca.head()
````

````python
# Merge to bring in YTD data
nasdaq100_ca = nasdaq100_ca.merge(
    price_change[["symbol", "ytd"]],
    on="symbol",
    how="left"
)

nasdaq100_ca.head()
````

````python
def classify_sector(name):
    prompt = f"""
    Classify the following company into one sector:
    Technology, Consumer Cyclical, Industrials, Utilities, Healthcare,
    Communication, Energy, Consumer Defensive, Real Estate, or Financial.

    Company: {name}

    Return ONLY the sector name.
    """

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content.strip()

# Apply classification
nasdaq100_ca["sector"] = nasdaq100_ca["name"].apply(classify_sector)
````

````python
summary_prompt = f"""
Here is Nasdaq-100 California company data with YTD performance:

{nasdaq100_ca.to_string()}

1. Pick the **two best performing sectors** overall.
2. Recommend **two or more companies** in each of those sectors.
3. Provide a concise explanation.

Return 1–3 paragraphs max.
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": summary_prompt}]
)

stock_recommendations = response.choices[0].message.content

stock_recommendations
````
