### The Data

The Google Trends data is available as a CSV file `ai_tools_comparison.csv`.
The data contains the number of worldwide searches for chatGpt, Gemini, and Microsoft Copilot over the past 12 months as of September 2024. 
<!--- Link to query: https://trends.google.com/trends/explore?q=chatgpt,gemini,%2Fg%2F11tsqm45vd&hl=en-US -->

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
trends = pd.read_csv('ai_tools_comparison.csv')

# Convert the 'week' column to datetime so we can treat it like a real timeline
trends['week'] = pd.to_datetime(trends['week'])

# Make sure the rows are in time order
trends = trends.sort_values('week').reset_index(drop=True)

# Preview the data structure 
display(trends.head())
display(trends.info())
```

```python
tools = ['chatgpt', 'gemini', 'microsoft_copilot']

# Calculate week-to-week percent change for each tool
growth_rates = trends[tools].pct_change() * 100

# Measure how "stable" each tool's growth is
std_dev = growth_rates.std(skipna=True)

# The tool with the lowest volatility in growth is considered most consistent
most_consistent_tool = std_dev.idxmin()

print("most_consistent_tool =", most_consistent_tool)
```

```python
min_idx = trends['chatgpt'].idxmin()  # row where ChatGPT is lowest
min_date = pd.to_datetime(trends.loc[min_idx, 'week'])  # get the actual week from that row
gpt_dip = min_date.strftime('%B %Y')
print(f"The worst month and year for ChatGPT: {gpt_dip}")
```

```python
monthly = trends.set_index('week')[tools].resample('MS').mean()
monthly_overall = monthly.mean(axis=1).dropna()

best_month_idx = monthly_overall.idxmax()
best_month = best_month_idx.strftime('%B')
print(f"Month with the highest interest: {best_month}")
```

```python
# Final answers

print("most_consistent_tool =", most_consistent_tool)
print("gpt_dip =", gpt_dip)
print("best_month =", best_month)
```
